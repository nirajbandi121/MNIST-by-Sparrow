```python
#SIN WAVE
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
import numpy as np
import math
```

```python
#Creat the real SIN Wave data
def get_real_data(batch_size):
  x=2*math.pi*torch.rand(batch_size,1)
  y=torch.sin(x)
  return torch.cat((x,y), dim=1)
```

```python
# Creating Police(Discriminator) and Thief(Generator)
# Generator
class Generator(nn.Module):
  def __init__ (self):
    super().__init__()
    self.net = nn.Sequential(
        nn.Linear(2, 16),
        nn.ReLU(),
        nn.Linear(16, 32),
        nn.ReLU(),
        nn.Linear(32,2)
      )
  def forward(self, noise):
    return self.net(noise)

#Discriminator

class Discriminator(nn.Module):
  def __init__ (self):
    super().__init__()
    self.net = nn.Sequential(
        nn.Linear(2, 32),
        nn.ReLU(),
        nn.Linear(32, 16),
        nn.ReLU(),
        nn.Linear(16 ,1),
        nn.Sigmoid()
        )
  def forward(self, data):
    return self.net(data)
generator = Generator()
discriminator = Discriminator()
```

```python
#Paramaters
loss_fn = nn.BCELoss()
opt_G = torch.optim.Adam(generator.parameters(), lr=0.001)
opt_D = torch.optim.Adam(discriminator.parameters(), lr=0.001)
```

```python
epochs = 100000
batch_size = 64

#Training Loop
for epoch in range(epochs+1):
  #Discriminator
  opt_D.zero_grad()

  #real data

  real_data = get_real_data(batch_size)
  real_labels = torch.ones(batch_size, 1)
  loss_real = loss_fn(discriminator(real_data), real_labels)

  #Fake data

  noise = torch.randn(batch_size, 2)
  fake_data = generator(noise).detach()
  fake_labels = torch.zeros(batch_size, 1)
  loss_fake = loss_fn(discriminator(fake_data), fake_labels)

  # Discriminator
  loss_D = loss_real+loss_fake
  loss_D.backward()
  opt_D.step()


    #Generator
  opt_G.zero_grad()

  #Fake data

  noise = torch.randn(batch_size, 2)
  fake_data = generator(noise)

  loss_G = loss_fn(discriminator(fake_data), real_labels)

  # Generator
  loss_G.backward()
  opt_G.step()

  # --- VISUALIZE PROGRESS ---
  if epoch % 1000 == 0:
        print(f"Epoch {epoch:4d} | Detective Loss: {loss_D.item():.3f} | Counterfeiter Loss: {loss_G.item():.3f}")

        with torch.no_grad():
            test_noise = torch.randn(200, 2)
            generated_points = generator(test_noise).numpy()
            real_points = get_real_data(200).numpy()

        plt.figure(figsize=(6, 4))
        plt.scatter(real_points[:, 0], real_points[:, 1], label="Real Sine Wave", alpha=0.5)
        plt.scatter(generated_points[:, 0], generated_points[:, 1], label="Fake (Generator)", color="red", alpha=0.7)
        plt.title(f"GAN Training Progress — Epoch {epoch}")
        plt.legend()
        plt.ylim(-1.5, 1.5)
        plt.show()
```
